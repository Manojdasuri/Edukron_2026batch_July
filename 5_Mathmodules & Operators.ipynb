```python
##Arithmetic Operators

##1.Write a program to perform addition, subtraction, multiplication, division, floor division, modulus, and exponentiation on two numbers.

a = 10
b = 3
print("addition:", a + b)
print("subtraction:", a - b)
print("multiplication:", a * b)
print("division:", a / b)
print("floor Division:", a // b)
print("modulus:", a % b)
print("exponentiation:", a ** b)
```

```python
##2.Swap two numbers without using a third variable.
a = 10
b = 15
print("Before:", a, b)
a, b = b, a
print("After:", a, b)
```

```python
##3.Find whether a number is even or odd using the modulus operator.
a=7
if a%2==0:
  print(a ,"is even")
else:
  print(a ,"is odd")
```

```python
##4.Reverse a three-digit number using arithmetic operators.
num=123
first=num//100
second=(num//10)%10
third=num%10
reverse =third*100 + second*10 + first
print(reverse)
```

```python
##5.Find the last digit and first digit of a given number.
num =12345
last =num%10
while num>10:
  num = num // 10
print("First Digit:", num)
print("Last Digit :", last)
```

```python
##Assignment Operators

##6.Demonstrate all assignment operators (=, +=, -=, *=, /=, //=, %= and **=) on a variable.
a=10
print("Initial Value:",a)
a+=5
print("After +=:",a)
a-=3
print("After -=:",a)
a*=2
print("After *=:",a)
a/=4
print("After /=:",a)
a//=2
print("After //=:",a)
a%=3
print("After %=:",a)
a**=3
print("After **=:",a)
```

```python
##7.Update a bank account balance using compound assignment operators.
balance = 1000
print("Bank Balance:", balance)
balance += 500
print("After Deposit:", balance)
balance -= 200
print("After Withdrawal:", balance)
```

```python
##21.Count how many bits are set to 1 in a number using bitwise operations.
num=13
count=0
while num>0:
    if num&1:
        count+=1
    num=num>>1
print("Set Bits:",count)
```

```python
##Membership Operators

##25.Check whether a character exists in a string.
name="manoj"
if "m" in name:
    print("Character exists in string")
else:
    print("Character does not exist in string")
```

```python
##26.Check whether a number exists in a list.
list=[1,2,3,4]
if 2 in list:
    print("Number exists in list")
else:
    print("Number does not exist in list")
```

```python
##27.Count how many vowels in a string belong to a predefined vowel list using in.
string="manoj"
vowels=["a","e","i","o","u"]
count=0
for char in string:
    if char in vowels:
        count+=1
print("Number of vowels:",count)
```

```python
##Mixed Operator Programs

##28.Build a simple calculator using arithmetic and comparison operators.
a=10
b=5
print("Addition:",a+b)
print("Subtraction:",a-b)
print("Multiplication:",a*b)
print("Division:",a/b)
print("Equal:",a==b)
print("Greater:",a>b)
print("Less:",a<b)
```

```python
###29.Create a grading system using comparison and logical operators.
marks=85
if marks>=90 and marks<=100:
    print("Grade A")
elif marks>=80 and marks<=89:
    print("Grade B")
elif marks>=70 and marks<=79:
    print("Grade C")
elif marks>=60 and marks<=69:
    print("Grade D")
else:
    print("Fail")
```

```python
##30.Build a menu-driven calculator that uses arithmetic, comparison, logical, assignment, membership, identity, and bitwise operators wherever applicable.
a=10
b=5
choice=1

print("1.Addition")
print("2.Subtraction")
print("3.Multiplication")
print("4.Division")

if choice==1:
    print("Addition:",a+b)
elif choice==2:
    print("Subtraction:",a-b)
elif choice==3:
    print("Multiplication:",a*b)
elif choice==4:
    print("Division:",a/b)

print("Greater:",a>b)
print("Logical:",a>b and b>0)

a+=b
print("Assignment:",a)

text="APPLE"
print("Membership:",'A' in text)

x=[1,2]
y=x
print("Identity:",x is y)

print("Bitwise AND:",10&5)
```
